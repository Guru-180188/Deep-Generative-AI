```python
import torch
import torch.nn as nn
from torchinfo import summary  # Importing summary tool

class SimpleNet(nn.Module):
    def __init__(self):
        super(SimpleNet, self).__init__()
        self.layer_0 = nn.Linear(8, 16)
        self.layer_1 = nn.Linear(16, 24)
        self.layer_2 = nn.Linear(24, 10)
        self.layer_3 = nn.Linear(10, 1)
        self.relu = nn.ReLU()
        self.tanh = nn.Tanh()

    def forward(self, x):
        x = self.layer_0(x)
        x = self.relu(x)
        x = self.layer_1(x)
        x = self.layer_2(x)
        x = self.tanh(x)
        x = self.layer_3(x)
        return x
```

```python
model = SimpleNet()
summary(model, input_size=(8, 8))  # (batch_size, input_features)
```
