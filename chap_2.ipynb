```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Define the function
def y_func(x1, x2):
    return np.sin(2 * np.pi * x1) * np.sin(2 * np.pi * x2)

# (a) Plot the true function y(x1, x2) over the grid
x1_grid = np.linspace(0, 1, 100)
x2_grid = np.linspace(0, 1, 100)
X1, X2 = np.meshgrid(x1_grid, x2_grid)
Y_true = y_func(X1, X2)

fig = plt.figure(figsize=(18, 5))

ax1 = fig.add_subplot(131, projection='3d')
ax1.plot_surface(X1, X2, Y_true, cmap='viridis', edgecolor='none')
ax1.set_title('(a) True function y(x1, x2)')
ax1.set_xlabel('x1')
ax1.set_ylabel('x2')
ax1.set_zlabel('y')

# (b) 100 data points with x1 observed, x2 unobserved (random) => high noise
N = 100
x1_b = np.random.uniform(0, 1, N)
x2_b = np.random.uniform(0, 1, N)  # unobserved, random across full range
noise_std = 0.3
y_b = y_func(x1_b, x2_b) + np.random.normal(0, noise_std, N)

ax2 = fig.add_subplot(132)
sc2 = ax2.scatter(x1_b, y_b, c='r', alpha=0.7)
ax2.set_title('(b) x2 unobserved (random), noisy y vs x1')
ax2.set_xlabel('x1')
ax2.set_ylabel('y')
ax2.grid(True)

# (c) 100 data points with x2 fixed to pi/2 / (2*pi) ~ 0.25 (scaled to [0,1])
x2_fixed = 0.25  # approx pi/2 / (2*pi)
x1_c = np.random.uniform(0, 1, N)
y_c = y_func(x1_c, x2_fixed) + np.random.normal(0, noise_std, N)

ax3 = fig.add_subplot(133)
sc3 = ax3.scatter(x1_c, y_c, c='b', alpha=0.7)
ax3.set_title('(c) x2 fixed = π/2/(2π), less noisy y vs x1')
ax3.set_xlabel('x1')
ax3.set_ylabel('y')
ax3.grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Outcomes (candies) and their "scores"
candies = ["Chocolate 🍫", "Lollipop 🍭", "Gummy 🐻"]
scores = [10, 5, 2]
probabilities = [2/6, 3/6, 1/6]

# Expected value calculation
expected_value = sum(s * p for s, p in zip(scores, probabilities))

# Plot
plt.bar(candies, scores, color="skyblue", edgecolor="black")
for i, (s, p) in enumerate(zip(scores, probabilities)):
    plt.text(i, s + 0.3, f"Prob={p:.2f}", ha="center")

plt.axhline(expected_value, color="red", linestyle="--", label=f"Expectation = {expected_value:.2f}")
plt.legend()
plt.title("Expectation Example: Candy Scores")
plt.ylabel("Score Value")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for Bernoulli distribution
p = 0.7  # probability of success (1)
n = 1000  # number of trials

# Generate Bernoulli samples
samples = np.random.binomial(n=1, p=p, size=n)

# Calculate mean (should be close to p)
mean = np.mean(samples)

# Plot the results
plt.bar([0, 1], [np.sum(samples == 0), np.sum(samples == 1)], color=['skyblue', 'orange'])
plt.xticks([0, 1], ['0 (Failure)', '1 (Success)'])
plt.title(f'Bernoulli Distribution (p={p})')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.text(0.5, max(np.sum(samples == 0), np.sum(samples == 1)) * 0.9, f'Mean = {mean:.2f}', ha='center', color='red')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Bernoulli distribution parameter
p = 0.7  # probability of success (1)

# Probability Mass Function (PMF) for Bernoulli
outcomes = [0, 1]
pmf = [1 - p, p]

plt.bar(outcomes, pmf, color=['skyblue', 'orange'])
plt.xticks(outcomes, ['0 (Failure)', '1 (Success)'])
plt.title(f'Bernoulli Distribution PMF (p={p})')
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.ylim(0, 1)
for i, prob in enumerate(pmf):
    plt.text(i, prob + 0.02, f'{prob:.2f}', ha='center', color='black')
plt.show()
```

## Bernoulli Distribution: Discrete Nature
The Bernoulli distribution is a discrete probability distribution. It describes the outcome of a single experiment that can result in just two possible values: success (1) or failure (0).
Unlike continuous distributions, the Bernoulli distribution only assigns probability to these two distinct outcomes.

## Binomial Distribution: Discrete Nature
The binomial distribution is a discrete probability distribution. It describes the number of successes in a fixed number of independent trials, each with the same probability of success.
Possible outcomes are whole numbers from 0 up to the total number of trials.


## What is PMF?
PMF stands for <b>Probability Mass Function</b>.
It is a function that gives the probability of each possible outcome for a discrete random variable.
For example, in the binomial or Bernoulli distribution, the PMF tells you the probability of getting each possible number of successes.
PMF is only defined for discrete distributions.

## Multinomial Distribution: Discrete Nature
The multinomial distribution is a discrete probability distribution.
It describes the probabilities of counts for each category in a fixed number of independent trials, where each trial results in one of several categories.
Possible outcomes are whole numbers (counts) for each category.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal

# Grid for plotting
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
pos = np.dstack((X, Y))

mean = [0, 0]

# (a) General covariance matrix
cov_general = [[2, 1.2], [1.2, 1]]
rv_general = multivariate_normal(mean, cov_general)
Z_general = rv_general.pdf(pos)

# (b) Diagonal covariance matrix
cov_diag = [[2, 0], [0, 1]]
rv_diag = multivariate_normal(mean, cov_diag)
Z_diag = rv_diag.pdf(pos)

# (c) Identity covariance matrix (proportional to identity)
cov_identity = [[1, 0], [0, 1]]
rv_identity = multivariate_normal(mean, cov_identity)
Z_identity = rv_identity.pdf(pos)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot (a) General covariance
axes[0].contour(X, Y, Z_general, levels=8, cmap='viridis')
axes[0].set_title('(a) General Covariance')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')

# Plot (b) Diagonal covariance
axes[1].contour(X, Y, Z_diag, levels=8, cmap='viridis')
axes[1].set_title('(b) Diagonal Covariance')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')

# Plot (c) Identity covariance
axes[2].contour(X, Y, Z_identity, levels=8, cmap='viridis')
axes[2].set_title('(c) Identity Covariance')
axes[2].set_xlabel('x')
axes[2].set_ylabel('y')

plt.tight_layout()
plt.show()
```

## Difference Between Covariance Matrix Types in 2D Gaussian Contours
- **General Covariance Matrix:** The off-diagonal elements are nonzero, so the contours are ellipses that can be rotated. This means the variables are correlated, and the principal axes of the ellipse are not aligned with the coordinate axes.
- **Diagonal Covariance Matrix:** Only the diagonal elements are nonzero, so the contours are ellipses aligned with the coordinate axes. The variables are uncorrelated, but can have different variances.
- **Identity Covariance Matrix:** The covariance matrix is proportional to the identity matrix, so all directions have the same variance. The contours are concentric circles, indicating no correlation and equal spread in all directions.

```python
# Conditional Gaussian demo with code (re-run after state reset)
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

np.random.seed(42)

# ---- Joint Gaussian parameters ----
mu_X = 170.0   # mean height (cm)
mu_Y = 70.0    # mean weight (kg)
sigma_X = 10.0 # std height
sigma_Y = 15.0 # std weight
rho = 0.8      # correlation

# Covariance matrix
Sigma = np.array([
    [sigma_X**2,          rho * sigma_X * sigma_Y],
    [rho * sigma_X * sigma_Y,  sigma_Y**2]
])

mu = np.array([mu_X, mu_Y])

# ---- Simulate data ----
n = 20000
samples = np.random.multivariate_normal(mean=mu, cov=Sigma, size=n)
X = samples[:, 0]
Y = samples[:, 1]

# ---- Condition on X = x0 ----
x0 = 180.0

# Theoretical conditional distribution: Y | X = x0
Sigma_xx = Sigma[0, 0]   # scalar
Sigma_xy = Sigma[0, 1]
Sigma_yx = Sigma[1, 0]
Sigma_yy = Sigma[1, 1]

mu_cond = mu_Y + (Sigma_yx / Sigma_xx) * (x0 - mu_X)
var_cond = Sigma_yy - (Sigma_yx * Sigma_xy) / Sigma_xx
std_cond = np.sqrt(var_cond)

# ---- Empirical approximation: take samples with X close to x0 ----
band = 0.5  # cm tolerance around x0
mask = (X >= x0 - band) & (X <= x0 + band)
Y_slice = Y[mask]

# ---- Print key numbers ----
print("=== Joint distribution parameters ===")
print(f"Mean (X, Y): ({mu_X:.1f}, {mu_Y:.1f})")
print(f"Std (X, Y):  ({sigma_X:.1f}, {sigma_Y:.1f})")
print(f"Correlation:  {rho:.2f}")
print("\n=== Conditioning on X = {:.1f} ===".format(x0))
print("Theoretical Y|X=x0 ~ Normal(mean={:.3f}, std={:.3f})".format(mu_cond, std_cond))
print("Empirical from simulated slice (|X - x0| <= {:.1f}):".format(band))
print("  Count in slice:", len(Y_slice))
print("  Empirical mean:", np.mean(Y_slice) if len(Y_slice) > 0 else np.nan)
print("  Empirical std: ", np.std(Y_slice, ddof=1) if len(Y_slice) > 1 else np.nan)

# ---- Plot 1: Joint scatter and the conditioning line ----
plt.figure()
plt.scatter(X[::20], Y[::20], s=8, alpha=0.4)  # downsample for clarity
plt.axvline(x0, linestyle="--", label=f"Condition X={x0:g}")
plt.title("Joint 2D Gaussian Samples (Height X, Weight Y)")
plt.xlabel("Height X (cm)")
plt.ylabel("Weight Y (kg)")
plt.legend()
plt.show()

# ---- Plot 2: Histogram of Y in the slice vs theoretical conditional PDF ----
plt.figure()
# Histogram
count, bins, _ = plt.hist(Y_slice, bins=30, density=True, alpha=0.6, label="Empirical Y | X≈x0")

# Theoretical PDF curve
y_grid = np.linspace(mu_cond - 4*std_cond, mu_cond + 4*std_cond, 400)
pdf = norm.pdf(y_grid, loc=mu_cond, scale=std_cond)
plt.plot(y_grid, pdf, linewidth=2, label="Theoretical N(mean, std)")

plt.title(f"Conditional Distribution: Y | X = {x0:g}")
plt.xlabel("Weight Y (kg)")
plt.ylabel("Density")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for linear Gaussian model
a = 2.0   # slope
b = 1.0   # intercept
sigma = 1.0  # standard deviation of Gaussian noise
N = 100

# Generate data
np.random.seed(42)
x = np.linspace(-5, 5, N)
noise = np.random.normal(0, sigma, N)
y = a * x + b + noise

# Fit linear regression for visualization
coeffs = np.polyfit(x, y, 1)
y_fit = np.polyval(coeffs, x)

# Plot data and regression line
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, label='Data (with Gaussian noise)')
plt.plot(x, a * x + b, 'r--', label='True model: y = ax + b')
plt.plot(x, y_fit, 'g-', label='Fitted regression line')
plt.title('Linear Gaussian Model')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import vonmises

# Parameters for von Mises distribution
mu = 0         # mean direction (radians)
kappa = 4      # concentration parameter (higher = more concentrated)
N = 1000       # number of samples

# Generate samples
samples = vonmises.rvs(kappa, loc=mu, size=N)

# Plot histogram of samples
plt.figure(figsize=(8, 5))
plt.hist(samples, bins=30, density=True, alpha=0.6, color='skyblue', label='Samples')

# Plot theoretical PDF
theta = np.linspace(-np.pi, np.pi, 200)
pdf = vonmises.pdf(theta, kappa, loc=mu)
plt.plot(theta, pdf, 'r-', lw=2, label='von Mises PDF')

plt.title('von Mises Distribution')
plt.xlabel('Angle (radians)')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal

# Parameters for 2D Gaussian
mean = [0, 0]
cov = [[2, 1], [1, 2]]

# Grid for plotting
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
pos = np.dstack((X, Y))

# Compute PDF
rv = multivariate_normal(mean, cov)
Z = rv.pdf(pos)

# Plot contour
plt.figure(figsize=(7, 6))
contour = plt.contourf(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(contour)
plt.title('2D Gaussian Distribution (Contour)')
plt.xlabel('x')
plt.ylabel('y')
plt.show()

# Plot 3D surface
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)
ax.set_title('2D Gaussian Distribution (Surface)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('Density')
plt.show()
```
