```python
import numpy as np
import matplotlib.pyplot as plt
Y = []
X = []
for x in range(10):
    y = np.exp(-x)
    X.append(x)
    Y.append(y)

plt.plot(X,Y)
plt.show()

print(f"the X values :: {X}")
print(f"the Y values :: {Y}")
```
